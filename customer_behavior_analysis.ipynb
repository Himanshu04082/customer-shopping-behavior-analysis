# Importing Library

```python
import pandas as pd
```

# Loading  dataset

```python

df = pd.read_csv('customer_shopping_behavior.csv')
```

# Displaying the first 5 rows of the dataset

```python
df.head()
```

# Checking dataset information such as columns, data types, and null values


```python
df.info()
```

# Checking missing values in each column

```python
df.isnull().sum()
```

# Filling missing values in 'Review Rating' using category-wise median

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

# Verifying if missing values have been handled

```python
df.isnull().sum()
```

# Converting column names to lowercase

```python
df.columns = df.columns.str.lower()

# Replacing spaces with underscore in column names
df.columns = df.columns.str.replace(' ','_')

# Renaming purchase amount column
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

# Displaying updated column names

```python
df.columns
```

# create a new column

```python
labels = ['Young Adults', 'Adult', 'Middle-Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

# Viewing age and corresponding age group columns

```python
df[['age', 'age_group']].head(10)
```

# create column purchase_frequency_days

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

# Displaying purchase frequency columns

```python
print(df[['purchase_frequency_days','frequency_of_purchases']].head(10))
```

# Checking discount applied and promo code used columns

```python
df[['discount_applied', 'promo_code_used']].head(10)
```

# Checking whether both columns contain same values or not

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

# Dropping promo_code_used column because it contains duplicate information

```python
df = df.drop(columns='promo_code_used', axis=1)
```

# Displaying final cleaned column names

```python
df.columns
```
